```python
#w5a16
def so_armstrong(x):
    if x < 0 :
        return False
    so_goc1 = x 
    so_goc2 = x 
    so_chu_so = 0 
    if so_goc2 == 0:
        so_chu_so = 1 
    else :
        while so_goc2 > 0 :
            so_chu_so += 1 
            so_goc2 //= 10
    
    tong = 0 
    so_goc2 = x 
    while so_goc2 > 0 :
        chu_so = so_goc2 % 10
        tong += chu_so ** so_chu_so 
        so_goc2 //= 10 
    return tong == so_goc1 
x = int(input())
print(so_armstrong(x))   

#cach 2 
def so_armstrong(x):
    s = str(x)
    n = len(s)
    tong = 0 
    for i in s :
        tong += int(i) ** n 
    return tong == x 
x = int(input())
print(so_armstrong(x))
```

```python
#w5a23
# a * (x **2) + b * x + c == 0
def quadratic_roots_count(a, b, c):
    delta = b ** 2 - 4 * a * c 
    if delta > 0 :
        return "2 nghiệm thực phân biệt"
    elif delta == 0 :
        return "1 nghiệm thực kép"
    else :
        return 0
a, b, c = map(int, input().split())
print(quadratic_roots_count(a, b, c))
```

```python
#w5a28 
import math
def goc_ABC(x1, x2, x3, y1, y2, y3):
    BAx = x1 - x2 
    BAy = y1 - y2 
    BCx = x3 - x2 
    BCy = y3 - y2 
    tich_vo_huong = BAx * BCx + BAy * BCy 
    tich_do_dai = (BAx ** 2 + BAy ** 2) ** 0.5 + (BCx ** 2 + BCy ** 2) ** 0.5
    cos = tich_vo_huong / tich_do_dai
    rad = math.acos(cos)
    degree = math.degrees(rad)
    return degree
x1, x2, x3, y1, y2, y3 = map(int,input().split())
print(goc_ABC(x1, x2, x3, y1, y2, y3))
```

```python
#w5a30
def C(n, k):
    if k < 0 or k > n :
        return 0
    giai_thua1 = 1
    giai_thua2 = 1 
    giai_thua3 = 1 
    for a in range (1, n + 1):
        giai_thua1 *= a
    for b in range (1, k + 1):
        giai_thua2 *= b
    for c in range(1, (n - k ) + 1):
        giai_thua3 *= c 
    return giai_thua1 / (giai_thua2 * giai_thua3)
n, k = map(int, input().split())
print(C(n, k))
```

```python
#w5a33
# #car_type = "4": giá mở cửa 10000
# car_type = "7": giá mở cửa 12000

# giá :
# 1 km đầu: 15000 / km
# từ km thứ 2 đến km thứ 10: 13000 / km
# từ km thứ 11 trở đi: 11000 / km

# phí chờ: 2000 mỗi phút
# phụ phí ban đêm: từ 22h đến trước 5h, cộng thêm 20%
# giảm giá: nếu đi trên 30 km thì giảm 10%

def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4 :
        mo_cua = 10000
    else:
        mo_cua = 12000
    
    if km <= 1 :
        gia_di_lai = km * 15000
    elif 2 <= km <= 10 :
        gia_di_lai = 1 * 15000 + (km - 1) * 13000
    else :
        gia_di_lai = 1 * 15000 + 9 * 13000 + (km - 10) * 11000

    tien_cho = wait_minutes * 2000 
    tong = mo_cua + gia_di_lai + tien_cho
    if start_hour >= 22 or start_hour < 5 :
        tong = tong * 1.2
    if km > 30 :
        tong = tong * 0.9 
    return tong
car_type, start_hour, km, wait_minutes = map(int, input().split())
print (smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

```python
#w5a34
def  sum_1_to_n(n):
    if n <= 0 :
        return 0 
    tong = 0
    for i in range (1, n + 1 ):
        s = str(i)
        if len(s) == len(set(s)):
            tong += i
    return tong 
n = int(input())
print(sum_1_to_n(n))


    
```
