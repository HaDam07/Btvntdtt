### Bài 3: Tìm giá trị trung bình của 1 list

```python
def average_lst(lst):
    lst_positive = []
    for i in lst :
        if i > 0 :
            lst_positive.append(lst)
        else :
            return "Invalid!"
        tong = 0 
        for i in lst_positive :
            tong += 0 
        return tong / len(lst_positive)
    return "Invalid"
assert average_lst([1, 2, 3, 4, 5]) == 3 
assert average_lst([-3, -5, -2, -9]) == "invalid"
assert average_lst([]) == "invalid"

print("Accepted")
```

```python
def average_lst(lst):
    tong = 0
    dem = 0
    for x in lst:
        if x > 0:
            tong += x
            dem += 1
    if dem == 0:
        return "invalid"
    return tong / dem

assert average_lst([1, 2, 3, 4, 5]) == 3
assert average_lst([-3, -5, -2, -9]) == "invalid"
assert average_lst([]) == "invalid"

print("Accepted")

```

### Bài 4: Kiểm tra một list số nguyên có phải là đối xứng hay không

```python
def ktra_doi_xung(lst):
    for i in range (len(lst) // 2):
        if lst[i] != lst[len(lst) - 1 - i]:
            return False
        return True
    return None 
assert ktra_doi_xung([1, 2, 3, 5 , 3, 2, 1]) == True
assert ktra_doi_xung([1, 3, 4 ,1, 2]) != True
assert ktra_doi_xung([]) == None

print("accepted!")
    
```

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau
3 4 6 7 1 9 0 0 23 0 0 34 67 89
0 0 0 0 3 4 6 7 1 9 23 34 67 89

```python
def day_so(lst):
    if lst:
        for x in lst.copy():
            if x == 0:
                lst.remove(x)
                lst.insert(0, x)
        return (lst)
    return None

assert day_so([1, 4, 0, 0, 3, 4, 4, 5, 0, 0]) == [0, 0, 0, 0, 1, 4, 3, 4, 4, 5]
print("accepted!")

```

```python
def move_zero(lst):
    if not lst:
        return None
    
    zeros = []
    others = []
    
    for x in lst:
        if x == 0:
            zeros.append(x)
        else:
            others.append(x)
    
    return zeros + others

assert move_zero([1, 4, 0, 0, 3, 4, 4, 5, 0, 0]) == [0, 0, 0, 0, 1, 4, 3, 4, 4, 5]
print("accepted!")

```

### Bài 6: Xóa tất cả các số giống nhau trong list, Các số khác giữ nguyên vị trí trước sau

```python
def day_so(day_so):
    ket_qua = []
    for i in day_so :
        if i not in ket_qua:
            ket_qua.append(i)
    return ket_qua

assert day_so([1, 2, 2, 3, 4, 4]) == [1, 2, 3, 4]
print("accepted!")


```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python
def day_so(n):
    tang = True
    giam = True
    for i in range (len(n) -1):
        if n[i] < n[i+1]:
            giam = False
        if n[i] > n[i+1]:
            tang = False 
    if tang:
        return "Tang"
    elif giam:
        return "Giam"
    else:
        return "Vo danh"
    
assert day_so([1, 3, 5, 6, 7, 8, 9]) == "Tang"
assert day_so([9, 8, 7, 6, 5, 4, 0]) == "Giam"
assert day_so([1, 2, 3, 7, 1, 0, 2]) == "Vo danh"

print("Pass all tests!")

            
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

```python
def ziczac(n):
    so_ziczac = True
    for i in range (len(n)-2):
        if not (n[i] < n[i+1] > n[i+2] or n[i] > n[i+1] < n[i+2]):
            so_ziczac = False 
    if so_ziczac:
        return "ziczac"
    return "not ziczac"
assert ziczac([1, 3, 2, 5, 2, 8, 7, 9, 2, 100]) == "ziczac"
assert ziczac([34, 2, 56, 7, 90, 20, 30, 7, 120]) == "ziczac"
assert ziczac([1, 5, 6, 7, 8, 3, 2, 2]) == "not ziczac"
print("Pass all tests!")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
def day_so(n, k):
    ket_qua = []
    for i, value in enumerate(n):
        if value == k :
            ket_qua.append(i)
    if ket_qua == []:
        return -1 
    else:
        return ket_qua 
n = list(map(int, input().split()))
k = int(input())
print(day_so(n, k))
```
