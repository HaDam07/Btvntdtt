```python
#w6a24 Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
#trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9
def day_so(n):
    zeros = []
    others = []
    for i in n :
        if i == 0 :
            zeros.append(i)
        else:
            others.append(i)
    return zeros + others
n = map(int, input().split())
print(day_so(n))
```

```python
#w6a26 Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối
def day_so(n):
    tong = 0 
    for i in  range (len(n) - 1):
        if n[i-1] < n[i] > n[i+1]:
            tong += n[i]
    return tong 
n = list(map(int, input().split()))
print(day_so(n))
```

```python
#w6a27 Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).nếu không có in ra -1
def day_so(n):
    result = []
    for i, value in enumerate(n):
        if value < x:
            result.append(i)
    if result == [] :
        return -1 
    else :
        return result
n = list(map(int, input().split()))
x = int(input())
print(day_so(n))
```

```python
#w6a28 Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm)
def day_so(n):
    tang = True 
    giam = True 
    for i in range (len(n) - 1):
        if n[i] > n[i+1]:
            tang = False 
        if n[i] < n[i+1]:
            giam = False
    if tang:
        return "Tang"
    elif giam: 
        return "Giam"
    else: 
        return "Vo danh"
n = list(map(int, input().split()))
print(day_so(n))
```
