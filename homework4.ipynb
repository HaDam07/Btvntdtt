```python
#w4a25
x = int(input("Nhap so: "))

if x == -1:
    print("invalid")
else:
    max = x
    min = x

    while True:
        x = int(input("Nhap so: "))
        if x == -1:
            break

        if x > max:
            max = x
        if x < min:
            min = x

    print(max, min)
```
