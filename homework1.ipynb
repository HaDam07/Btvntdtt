```python
#w1a6
a, b, c = map(int, input().split())
if a + b < c and a + c < b and b + c < a :
    print("Khong phai 3 canh cua tam giac ")
else :
    ncv = (a + b + c) / 2
    s = (ncv * (ncv - a) * (ncv - b) * (ncv - c)) ** 0.5 
    print(f"{s:.1f}")
```
