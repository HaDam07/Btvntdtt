```python
#w6a1
day_so = list(map(int, input().split()))
ket_qua = []
for i in day_so :
    if not i in ket_qua:
        ket_qua.append(i)
print(ket_qua)
```

```python
#w6a5
n = list(map(int, input().split()))
positives = 0 
negatives = 0 
zeros = 0

for i in n :
    if i > 0 :
        positives += 1 
    elif i < 0: 
        negatives += 1 
    else :
        zeros += 1
ket_qua = {
'positives' : positives,
'negatives' : negatives,
'zeros' : zeros
}
print(ket_qua)
```
