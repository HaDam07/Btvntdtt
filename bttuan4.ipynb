Hãy viết một chương trình nhận vào một số nguyên dương n và một dãy gồm n số
nguyên. Nếu n số nguyên đó có số 42 thì in ra màn hình dòng chữ "I′ve found the
meaning of life!", ngược lại in ra dòng chữ "It′s a joke!".

```python
#w4a5
n = int(input())
so = list(map(int, input().split()))
if n >= 2 and len(so) == n:
    tim_thay = False
    for i in so :
        if i == 42 :
            tim_thay = True
            break
    if tim_thay:
        print("I've found the meaning of life!")
    else:
        print("It's a joke!")
```

```python
#w4a7
while True:
    n = int(input())
    if n>=2:
        break
la_so_nguyen_to = True
for i in range(2, n):
    if n % i == 0:
        la_so_nguyen_to = False
        break
if la_so_nguyen_to:
    print(n)
else:
    uoc_nguyen_to = []
    for j in range (2, n):
        la_so_nguyen_to_2 = True
        for k in range(2,j):
            if j % k == 0: 
                la_so_nguyen_to_2 = False
                break
        if la_so_nguyen_to_2:
            if n % j == 0:
                uoc_nguyen_to.append(j)
print(max(uoc_nguyen_to))
```

```python
#w4a13
def tong(n): 
  tong = 0
  for i in range (1, n):
    if n % i == 0 :
      tong += i 
  return tong 
a, b = map(int, input(). split())
if tong(a) == b and tong(b) == a :
  print("true")
else :
  print("false")
```
