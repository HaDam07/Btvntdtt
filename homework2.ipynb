```python
#bai 3
def tam_giac () :
   a, b, c = map(float, input().split())
   if a + b <= c or a + c <= b or b + c <= a : 
    return "khong phai 3 canh cua mot tam giac" 
   if a == b == c :
    loai = "tam giac deu"
   elif a == b or b == c or c == a : 
    loai = "tam giac can"
   elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2 : 
    loai = "tam giac vuong"
   else : 
    loai = "tam giac thuong" 
   P = a + b + c
   p = P / 2
   S = (p * ( p - a ) * ( p - b )* ( p - c )) ** 0.5

   return f"{loai}, chu vi P = {P}, dien tich S = {S}"
print(tam_giac()) 
```

```python
#bai 5
def giai_pt_bac2(a , b ,c):
    delta = b**2 - 4*a*c
    if delta < 0:
        print(" vo nghiem")
    elif delta == 0:
        x = (-b)/2*a
        print(" co nghiem kep: ", x)
    elif delta > 0:
        x1 = (-b + delta**(1/2))/2*a
        x2 = (-b - delta**(1/2))/2*a
        print(" 2 nghiem phan biet: ",x1 , x2)
a , b , c = map(float , input().split())
giai_pt_bac2(a , b ,c)
```
