```python
#W5A3 Viết hàm kiểm tra một số có phải là số nguyên tố hay không?
def so_nguyen_to(x):
    if x < 2 :
        return False
    for i in range (2, int(x ** 0.5 ) + 1):
        if x % i == 0 :
         return False 
    else : 
        return True
x = int(input())
print(so_nguyen_to(x))



    
```

```python
#w5a4
def so_hoan_hao(n):
    tong = 0 
    for i in range (1, n):
        if n % i == 0: 
         tong +=  i 
    if tong == n :
        return True
    else :
        return False
n = int(input())
print(so_hoan_hao(n))
```

```python
#w5a7
def casio(num1, num2, operator):
    if operator == "+":
        return (round(num1 + num2, 2))
    elif operator == "-":
        return (round(num1 - num2, 2))
    elif operator == "/":
        return (round(num1 / num2, 2))
    if operator == "*":
        return (round(num1 * num2, 2))
num1, num2 = map(int, input().split())
operator = input()
print(casio(num1, num2, operator))
```

```python
#w5a8
def hamming_distance(x, y):
    z = x ^ y 
    dem = 0 
    while z > 0 :
        dem += z % 2
        z //= 2 
    return dem 
x, y = map(int, input().split())
print(hamming_distance(x, y))
```

```python
#w5a9
def tong_cac_chu_so(n):
    tong = 0 
    while n > 0: 
        tong += n % 10 
        n //= 10 
    return tong
n = int(input())
print(tong_cac_chu_so(n))
```

```python
#w5a10
def dang_cau(a, b): 
    anh_xa_ab ={}
    anh_xa_ba ={}
    for i in range (len(a)):
        x = a[i]
        y = b[i]
        if x in anh_xa_ab:
            if anh_xa_ab[x] != y :
                return False 
        else :
            anh_xa_ab[x] = y
        if y in anh_xa_ba:
            if anh_xa_ba[y] != x:
                return False
        else:
            anh_xa_ba[y] = x 
    return True
a, b = input().split()
print(dang_cau(a, b))
```

```python
#w5a12
#*phan tich de bai : dividend = divisor * q + r ==> q la thuong, r la du 
def divide_with_remainder(dividend, divisor):
    if divisor < 0:
        return None
    q = dividend // divisor 
    r = dividend % divisor 
    return(q, r)
dividend, divisor = map(int, input().split())
print(divide_with_remainder(dividend, divisor))
```
