```python
#bai 4
h = int(input())

for i in range(1, h + 1):
    if i == 1:
        print(" " * (h - 1) + "*")
    elif i == h:
        print("*" * (2 * h - 1))
    else:
        print(" " * (h - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
#bai14
x = int(input("Nhap so: "))

if x == -1:
    print("invalid")
else:
    max = x
    min = x

    while True:
        x = int(input("Nhap so: "))
        if x == -1:
            break

        if x > max:
            max = x
        if x < min:
            min = x

    print(max, min)
```

```python
#bai16
n = int(input())

chan = 0
le = 0

while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        chan += 1
    else:
        le += 1
    n //= 10

print(chan, le)
```
