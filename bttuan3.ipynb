```python
#w3a16
n = float(input())

t = int(n)

if n >= 0 or n == t:
    down = t
else:
    down = t - 1

if n <= 0 or n == t:
    up = t
else:
    up = t + 1

if n - down < up - n:
    nearest = down
else:
    nearest = up

print(up, down, nearest)
```
